````python
import pandas as pd
import matplotlib.pyplot as plt
````

````python
# 1. Percentage of passengers survived
survival_percentage = df["Survived"].mean() * 100
survival_percentage
````

````python
# 2. Survival rate by passenger class
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100
survival_by_class
````

````python
# 3. Survival rate by gender
survival_by_gender = df.groupby("Sex")["Survived"].mean() * 100
survival_by_gender
````

````python
# 4. Average fare by passenger class
average_fare_by_class = df.groupby("Pclass")["Fare"].mean()
average_fare_by_class
````

````python
# 5. Average age of survivors and non-survivors
average_age = df.groupby("Survived")["Age"].mean()
average_age
````

````python
# 6. Top 10 passengers by fare
top_10_fares = df.sort_values("Fare", ascending=False).head(10)
top_10_fares[["Name","Sex","Fare","Survived"]]
````

````python
# 7. Family size
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df.head(3)
````

````python
# Passengers travelling alone
alone = df[df["FamilySize"] == 1]
alone[["Name","Sex","Fare","Survived"]]
````

````python
# 8. Families larger than four
large_families = df[df["FamilySize"] > 4]
large_families[["Name","Sex","Fare","Survived"]]
````

````python
# 9. Extract passenger titles
df["Title"] = df["Name"].str.extract(r",\s*([^.]*)\.")
df.head(3)
````

````python
title_counts = df["Title"].value_counts()
title_counts
````

````python
# 10. Correlation between age and fare
age_fare_correlation = df["Age"].corr(df["Fare"])
age_fare_correlation
````

````python
# 11. Average fare by embarkation port
average_fare_by_port = df.groupby("Embarked")["Fare"].mean()
average_fare_by_port
````

````python
# 12. Number of passengers in each class
passengers_by_class = df["Pclass"].value_counts().sort_index()
passengers_by_class
````

````python
# 13. Missing age
missing_age = df["Age"].isna().sum()
missing_age
````

````python
# 14. Missing cabin
missing_cabin = df["Cabin"].isna().sum()
missing_cabin
````

````python
# 15. Percentage of females survived
female_survival = (df[df["Sex"] == "female"]["Survived"].mean() * 100)
female_survival
````

````python
# 16. Percentage of males survived
male_survival = (df[df["Sex"] == "male"]["Survived"].mean() * 100)
male_survival
````

````python
# 17. Age groups
bins = [0, 18, 30, 50, 100]
labels = ["0-18", "19-30", "31-50", "51+"]

df["AgeGroup"] = pd.cut(df["Age"],bins=bins,labels=labels)
df.head(3)
````

````python
age_group_survival = (df.groupby("AgeGroup")["Survived"].mean() * 100)
age_group_survival
````

````python
age_group_survival.plot(kind="bar",figsize=(4,2))
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")
plt.title("Survival Rate by Age Group")
plt.show()
````

````python
age_group_survival.plot(kind="pie",figsize=(6,4))
plt.title("Survival Rate by Age Group")
plt.show()
````

````python
# 18. First-class survival
first_class_survival = (df[df["Pclass"] == 1]["Survived"].mean() * 100)
first_class_survival
````

````python
# 19. Passengers paying below average fare
average_fare = df["Fare"].mean()

below_average_fare = df[df["Fare"] < average_fare]
below_average_fare[["Name","Sex","Fare","Survived"]]
````

````python
# 20. Passengers paying more than twice average fare
above_twice_average = df[df["Fare"] > 2 * average_fare]
above_twice_average[["Name","Sex","Fare","Survived"]]
````
