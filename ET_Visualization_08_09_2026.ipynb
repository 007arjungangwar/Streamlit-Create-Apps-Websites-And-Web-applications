```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Titanic-Dataset.csv")
df.head(3)
```

```python
df.shape
```

## **Question1: How does the average fare vary across passenger classes?**

```python
df.groupby("Pclass")["Fare"].mean()
```

```python
avg_fair_by_class=df.groupby("Pclass")["Fare"].mean()
avg_fair_by_class
```

```python
type(avg_fair_by_class)
```

### **Bar**

```python
import matplotlib.pyplot as plt
```

```python
from matplotlib import legend
avg_fair_by_class.plot.bar(x="Pclass",y="Fare",figsize=(4,2),legend=True,color="red",alpha=0.9,grid=True,edgecolor="black")
plt.show()
```

```python
df.groupby("Pclass")["Fare"].mean().plot(kind="bar",figsize=(4,2),color=['r','g','b'],legend=True) #color=['blue','black','red']
plt.title("The average fare vary across passenger classes")
plt.xlabel("Pessanger class")
plt.ylabel("fair in $ ")
plt.grid(True)
plt.show()
```

```python
df.groupby("Pclass")["Fare"].mean().plot(kind="bar",figsize=(4,2),cmap="RdYlGn",alpha=0.5,legend=True)
plt.title("The average fare vary across passenger classes")
plt.xlabel("Pessanger class")
plt.ylabel("fair in $ ")
plt.show()
```

### **Barh**

```python
df.groupby("Pclass")["Fare"].mean().plot(kind="barh",figsize=(4,2),color=['r','g','b'],legend=True)
plt.title("The average fare vary across passenger classes")
plt.xlabel("Pessanger class")
plt.ylabel("fair in $ ")
plt.grid(True)
plt.show()
```

### **Line**

```python
df.groupby("Pclass")["Fare"].mean().plot(kind="line",figsize=(5,3),marker="o",markersize=15,linewidth=4,color="green",linestyle="dashed")
plt.title("The average fare vary across passenger classes")
plt.xlabel("Pessanger class")
plt.ylabel("fair in $ ")
plt.grid(True)
plt.show()
```

### **Pie**

```python
df.groupby("Pclass")["Fare"].mean().plot(kind="pie",center=(9,5),figsize=(5,3),shadow=True,explode=(0.1,0.1,0.1),legend=True,autopct='%.2f%%')
plt.title("The average fare vary across passenger classes")
plt.legend(['class1','class2','class3'],title="Pessanger class",alignment='left')
plt.xlabel("Pessanger class")
plt.ylabel("fair in $ ")
plt.grid(True)
plt.show()
```

```python
df.groupby("Pclass")["Fare"].mean().plot(kind="pie",figsize=(5,3),shadow=True,explode=(0.1,0.1,0.1),legend=True,center=(0.5, 0.5),labeldistance=0.5,radius=1)
plt.title("The average fare vary across passenger classes")
plt.xlabel("Pessanger class")
plt.ylabel("fair in $ ")
plt.grid(True)
plt.show()
```

### **Scatter**

```python
avg_fare_by_class = df.groupby("Pclass")["Fare"].mean()
plt.figure(figsize=(5,3))
plt.scatter(x=avg_fare_by_class.index, y=avg_fare_by_class.values, marker="o", s=150, color="green")
plt.title("The average fare vary across passenger classes")
plt.xlabel("Pessanger class")
plt.ylabel("fair in $ ")
plt.grid(True)
plt.show()
```
